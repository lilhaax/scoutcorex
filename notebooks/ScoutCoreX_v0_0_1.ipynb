Goal

Retrieve scientific papers from arXiv using a natural language query.

Current version only focuses on paper retrieval.

Import required libraries

```python
import requests
import xml.etree.ElementTree as ET
```

Build the arXiv API URL

```python
base_url = "https://export.arxiv.org/api/query"
params = {
    'search_query': 'all:machine learning',
    'start':0,
    'max_results':5
}
```

```python
response = requests.get(base_url, params=params)
print(response.url)
```

```python
class ArxivPaper:
  def __init__(self, entry):
    self.entry = entry
    self.ns = {'atom': "http://www.w3.org/2005/Atom"}

  def get_title(self):
    return self.entry.find("atom:title",self.ns).text.strip()

  def get_authors(self):
    authors = []
    for author in self.entry.findall('atom:author',self.ns):
      name = author.find('atom:name', self.ns).text
      authors.append(name)

    return ", ".join(authors)

  def get_published(self):
    return self.entry.find('atom:published', self.ns).text[:10]

  def get_link(self):
    return self.entry.find('atom:id', self.ns).text

  def get_abstract(self):
    return self.entry.find('atom:summary', self.ns).text.strip()

  def display(self, number):
        print("=" * 50)
        print(f"Paper {number}")
        print("=" * 50)

        print("\nTitle:")
        print(self.get_title())

        print("\nAuthors:")
        print(self.get_authors())

        print("\nPublished:")
        print(self.get_published())

        print("\nLink:")
        print(self.get_link())

        print("\nAbstract:")
        print(self.get_abstract())

        print("\n")
```

```python
root = ET.fromstring(response.text)

ns = {'atom': "http://www.w3.org/2005/Atom"}

papers = root.findall('atom:entry', ns)

for i,paper in enumerate(papers, start = 1):
  arxiv_paper = ArxivPaper(paper)
  arxiv_paper.display(i)
```
